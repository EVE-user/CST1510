*Walkthrough: `02 - Methods`*

### D5. Add a method

1. Add a method `summary(self)` to the class below. It **returns** a string like `R-001: 87.0/100.0`,
   built from `self.label`, `self.value` and `self.limit`.
2. Create a record and print the result of calling `.summary()` on it.

**Expected output:** `R-001: 87.0/100.0`

````python
# D5
class Record:
    def __init__(self, label, value, limit):
        """Create a record with a label, value and limit."""
        self.label = label
        self.value = value
        self.limit = limit

    def summary(self):
        """Return the record's information in the provided format."""
        return f"{self.label} : {self.value}/{self.limit}"

# Create a record and display its summary.
r = Record("R-001", 87.0, 100.0)
print(r.summary())
 
````

### D6. `self` is whichever object made the call

1. Copy your class from D5 and add a method `status(self)` that returns `"OVER LIMIT"` (100% or more),
   `"WARNING"` (90% or more) or `"OK"`, worked out from `self.value` and `self.limit`.
2. Create three records and print each one's `.status()`:

| Record | Expected output |
|---|---|
| `Record("R-001", 87.0, 100.0)` | `OK` |
| `Record("R-002", 120.0, 100.0)` | `OVER LIMIT` |
| `Record("R-003", 95.0, 100.0)` | `WARNING` |

````python
# D6
class Record:
    def __init__(self, label, value, limit):
        """Create a record with a label, value and limit."""
        self.label = label
        self.value = value
        self.limit = limit

    def percent(self):
        """Return the value as a percentage of the limit."""
        return (self.value) / (self.limit)*100
    def status(self):
        """Return status of the percent"""
        
        y=self.percent() # Stores the percentage in y.
        if y >= 100: # Compare the percentage to the limit and provide status.
           return "OVER LIMIT"
        elif y>= 90:
            return "WARNING"
        else:
            return "OK"

# Create a record and displays its status.
r = Record("R-001", 87.0, 100.0)
print(r.status())
````

### D7. One method calls another

1. Copy your class from D6 and add a method `percent(self)` that returns `(self.value / self.limit) * 100`.
2. Change `status(self)` so it uses `self.percent()` instead of working out the percentage itself.
3. Create `Record("R-001", 87.0, 100.0)` and print its `.percent()` and `.status()`.

**Expected output:** `87.0`, then `OK`

````python
# D7
# Define a class to store record information.
class Record:
    def __init__(self, label, value, limit):
        """Create a record with a label, value and limit."""
        self.label = label
        self.value = value
        self.limit = limit

    def percent(self):
        """Return the value as a percentage of the limit."""
        return (self.value)/(self.limit)*100
    
    def status(self):
        """Return status based on the percentage."""
        
        y=self.percent() # Store the percentage in y.
        if y >= 100:
           return "OVER LIMIT"
        elif y >= 90:
            return "WARNING"
        else:
            return "OK"

# Create a record and display its percent and status.
r = Record("R-001", 87.0, 100.0)
print(f"Percent:{r.percent()} , Status:{r.status()}.")
````

### D8. A method can change its own object

1. Define a class `Checker`. In `__init__(self)`, set `self.over_limit_count = 0`.
2. Add a method `check(self, value, limit)` that adds 1 to `self.over_limit_count` when `value` is
   greater than `limit`.
3. Create one `Checker`, call `.check()` three times with `(87, 100)`, `(120, 100)` and `(130, 100)`,
   then print `over_limit_count`.

**Expected output:** `2`

````python
# D8
class Checker:
    def __init__(self):
        """Start the count at zero."""
        self.over_limit_count = 0 # Start the count at zero.
    def check(self, value, limit):
        """Check if the value is greater than the limit."""
        if value > limit:
            self.over_limit_count += 1 # Increase the count by one.

# Create a Checker and call check() three times.
c = Checker()
c.check(87, 100)
c.check(120, 100) 
c.check(130, 100)

# Display the final count.
print(c.over_limit_count)
````

### D11. Load objects from a file

`records_sample.txt` (in this `LAB` folder) has one record per line, like `R-001,87,100`.

1. Write `load_records(path)` that reads the file and returns a list of `Record` objects
   (convert `value` and `limit` with `float()`).
2. Loop over the list and print each record's label and status.

**Expected output:** the same three lines as D9.

````python
# D11
def load_records(path):
    """Load the records from a file."""
    records = [] # Create an empty list to store the records.

    # Open the file and go through each line.
    with open(path) as f:
        for line in f:
            line = line.strip() # Remove the white space from the beginning and end of the line.
            label, value, limit = line.split(",")
            records.append(Record(label, float(value), float(limit)))
    return records

# Load records from the given path.
records = load_records("records_sample.txt")

# Loop over the records and display the label and status of the records.
for record in records:
 print(record.label, record.status())
````

### C2. An account that locks

Write a class `Account` with a `username` and a count of failed logins that starts at 0.
It has two methods:

- `fail(self)`: adds 1 to the failed logins
- `locked(self)`: returns `True` if there have been 3 or more failed logins

**Test it:**

| Code | Expected output |
|---|---|
| an account for `"anna"` with `.fail()` called 3 times, then `print(anna.locked())` | `True` |
| an account for `"ben"` with `.fail()` called once, then `print(ben.locked())` | `False` |

````python
# C2
class Account:
    def __init__(self, username):
        """Create an account with a username and zero count of failed attempts."""
        self.username = username
        self.count = 0
    
    def fail(self):
        """Add the number of failed attempts by one."""
        self.count += 1

    def locked(self):
        """Return True if the failed attempts are at least three."""
        return self.count >= 3

# Create an account for anna and call fail() three times.
anna = Account("anna")
anna.fail()
anna.fail()
anna.fail()
print(anna.locked())

# Create an account for ben and call fail() once.
ben = Account("ben")
ben.fail()
print(ben.locked())
                
````
